```python
import pandas as pd

df = pd.read_csv("../runs/detect/low/vessels/yolo26n_(baseline,UNIFORM)2/results.csv")
df.columns = df.columns.str.strip()

best = df.sort_values(
    by=["metrics/mAP50-95(B)", "metrics/mAP50(B)"],
    ascending=False
).iloc[0]

print(best.to_string())
```
